# 01 – Eksploracja danych (EDA)
Zbiór: Kaggle *Credit Card Fraud Detection* (`mlg-ulb/creditcardfraud`) → `data/raw/creditcard.csv`.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, matplotlib.pyplot as plt
from anomaly.data import load_raw, temporal_split, TARGET
df = load_raw('../data/raw/creditcard.csv')   # na start: ../data/raw/synthetic.csv
df.shape, df[TARGET].value_counts()
```

```python
print('Odsetek oszustw: %.4f%%' % (100*df[TARGET].mean()))
df.describe().T.head(10)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 3.5))
df[df[TARGET]==0]['Amount'].clip(upper=500).hist(bins=50, ax=ax[0]); ax[0].set_title('Kwota: normalne (<=500)')
df[df[TARGET]==1]['Amount'].clip(upper=500).hist(bins=50, ax=ax[1]); ax[1].set_title('Kwota: oszustwa (<=500)')
plt.tight_layout(); plt.savefig('../docs/eda_amount.png', dpi=150)
```

```python
tr, va, te = temporal_split(df)
pd.DataFrame({'n':[len(tr),len(va),len(te)], 'oszustwa':[tr[TARGET].sum(),va[TARGET].sum(),te[TARGET].sum()]},
             index=['train','val','test'])
```
